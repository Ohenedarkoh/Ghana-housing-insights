TRAIN-TEST SPLIT

```python
# the goal is to see how our model will perform on new, unseen data
# training set teaches the model
# testing set evaluates how well the model generalizes

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)


print(f"Training data shape is ",x_train.shape,"and testing data shape is", x_test.shape)



# log transformation for 'Price'
import numpy as np
y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)

print("Training data shape is", x_train.shape,"and testing data shape is",x_test.shape)
```

FEATURE SCALING(FEATURE ENGINEERING)

```python
# feature scaling(standardization) , is a type of preprocessing(part of feature engineering)
# makes the model more stable and improves accuracy

# ml models like linear regression assume that all features are on a similar scale 
# if one feature has a value of 5000 and another has a value of 20, the model will give importance to the larger scale feature
# we want to prevent the imbalance, we use StandardScaler, which transforms each feature


# we fit the scaler(transformers) only on training set

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(x_train)




x_train_scaled= scaler.transform(x_train)

x_test_scaled = scaler.transform(x_test)
```

```python
#imputation
from sklearn.impute import SimpleImputer

# we use median because it is robust against outliers
imputer = SimpleImputer(strategy ='median')

#fit on training data only 
x_train_imputed= imputer.fit_transform(x_train)
x_test_imputed = imputer.transform(x_test)

#scale imputed data
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train_imputed)
x_test_scaled = scaler.transform(x_test_imputed)


```

MODEL TRAINING

```python
#import model
# using random forest search cv will help us find a better combination of parameters for the model
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint


#initialize model
base_model = RandomForestRegressor(random_state=42, n_jobs=-1)

param_dist = {
    'n_estimators': randint(100, 500), # tress in forest
    'max_features': ['sqrt', 'log2', None], # features to consider
    'max_depth': randint(10, 50), # maximum number of levels in tree
    'min_samples_split': randint(2, 10), # minimum number of samples required to split an internal node
    'min_samples_leaf': randint(1, 5), # Minimum number of samples required to be at a leaf node
    'bootstrap': [True, False] # Method for sampling data points (with or without replacement)
}

# initialize RandomSearchCV
random_search = RandomizedSearchCV(
    estimator=base_model,    
    param_distributions=param_dist,
    n_iter=50,    # n_iter :number if paranmeters settings that are samplified
    cv=5,    # number of folds for cross-validation
    scoring='r2',    # metrci to optimize
    verbose=2,   # Controls the verbosity: the higher, the more messages.
    random_state=42,    #For reproducibility.
    n_jobs=-1     # use all available cores
)

# fit randomizedsearchcv to find the best hyperparameters
random_search.fit(x_train_scaled,y_train_log)


print("Best parameters found:", random_search.best_params_)

print("Best R² score from cross-validation:", random_search.best_score_)


model = random_search.best_estimator_


```

```python
print("Best parameters found:", random_search.best_params_)

print("Best R² score from cross-validation:", random_search.best_score_)
```

```python
#train model using scaled training data
model.fit(x_train_scaled,y_train_log)

#make predictions on test set to see how well it performs on unseen data
y_pred_log= model.predict(x_test_scaled)

# inverse transform prediction for metrics

y_pred = np.expm1(y_pred_log)
```

```python
#evaluating models performance using regression metrics
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
import pandas as pd
import numpy as np
```

```python
#r2 tells how much of the variation in price is explained by the model( close to 1 is better)
r2 = r2_score(y_test,y_pred)
```

```python
#RMSE gives an idea of error magnitude in the same unit as price
rmse = np.sqrt(mean_squared_error(y_test,y_pred))
```

```python
#MAE is the average absolute error in predictions
mae = mean_absolute_error(y_test,y_pred)
```

model performance

```python
print("R² score is", r2)
print("RMSE is", rmse)
print("MAE is", mae)
```
