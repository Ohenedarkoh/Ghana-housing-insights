```python
housing['Price']=housing['Price'].str.extract(r'(\d+,?\d+)').replace(',','',regex=True).astype(float)

#using regex to extract the numeric values from the price column's values
```

```python
for col in ['Bedrooms','Bathrooms','Garage']:
    housing[col]= pd.to_numeric(housing[col], errors='coerce').astype('Int64')
```

```python
housing.dtypes
```

```python
# Split amenities into individual columns
amenities = housing['Amenities'].str.split(', ', expand=True)

# Get unique amenities and create binary columns
all_amenities = set()
for i in range(amenities.shape[1]):
    all_amenities.update(amenities[i].dropna().unique())

for amenity in all_amenities:
    if pd.notna(amenity):  # Skip NaN values
        housing[f'Amenity_{amenity.strip()}'] = housing['Amenities'].fillna('').str.contains(amenity, regex=False).astype(int)



# Drop original column
# housing = housing.drop('Amenities', axis=1) not neccesary!
```

```python
housing.columns
```

```python
housing.head(2)
```

```python
housing = housing.drop([
    'Title',
    'URL',
    'Commission',
    'Amenities',
    'Address',
    'Reference',
    'Description',
],
axis= 1,
)


# dropped irrelevant columns
```

```python
housing.sort_values('Price', axis= 0, ascending=True, inplace= False).head(10)
```

```python
housing.isna().sum()
```

```python
housing = housing.dropna(subset=['Price','Bathrooms'])

# removed observations with no price and 0 bathrooms
```

```python
housing.isna().sum()
```

```python
housing['Furnished'].unique()
```

```python
housing['Furnished'] = housing['Furnished'].str.lower()
housing['Furnished'] = housing['Furnished'].map({"furnished":1}).fillna(0).astype(int)

#encoding furnished
```

```python
housing['Garage'].unique()

# inspected garage feature and realised it is a float, so changed it to int to remove decimals
```

```python
housing['Garage'] = housing['Garage'].fillna(0).astype(int)
```

```python
housing.info()
```

```python
columns = [
    "Bedrooms",
    "Bathrooms",
    "Garage",
]  # Determining the distribution of the numeric features

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for idx, col in enumerate(columns):
    housing[col].value_counts().sort_index().plot(
        kind="bar", ax=axes[idx], rot=0, grid=True
    )
    axes[idx].set_xlabel(f"Number of {col.lower()}")
    axes[idx].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
housing['Bedrooms'] = housing['Bedrooms'].clip(
    upper = 10 , lower = 1
) # Triming the data with some upper and lower bounds to have a standard distribution

housing['Bathrooms'] = housing['Bathrooms'].clip(upper = 10)
housing['Garage'] = housing['Garage'].clip(upper=10)
```

```python
housing['Price'].hist(bins=50, figsize=(4,4))

# Realize how price is skewed, linear models tend to underperform when the target is imbalanced, normal price in general is always skewed, a conventional way to handle price is log transformation
plt.show()
```

```python
housing['LogPrice'] = np.log(
    housing['Price']
)

# In an instance where there are zero prices, and you choose to keep them, to prevent a log(0) error, you find np.log(x) rather to handle 0 values
```

```python
housing['LogPrice'].hist(bins = 25 ,figsize =(4,4))

# with this you find the anti-log of the predictions if you want to express it in terms of real prices, thus e^pred
# after transformation a close bell shapped distribution is achieved
```

```python
cols = [
    "Amenity_Air conditioning",
    "Amenity_Fans",
    "Amenity_Gym",
    "Amenity_Refrigerator",
    "Amenity_Washing machine",
    "Amenity_Alarm system",
    "Amenity_Fully fenced",
    "Amenity_Security service",
    "Amenity_Intercom",
    "Amenity_Balcony",
    "Amenity_Garden",
    "Amenity_Swimming pool",
    "Amenity_Roof terrace",
    "Amenity_Outdoor entertaining area",
    "Amenity_Garage",
    "Amenity_Carport",
    "Amenity_Water reservoir",
    "Amenity_Generator",
    "Amenity_Cooker",
    "Amenity_Dishwasher",
    "Amenity_Oven",
    "Amenity_Microwave",
    "Amenity_Kitchen cabinets",
    "Amenity_Kitchen",
    "Amenity_Pay TV access",
    "Amenity_Dstv",
    "Amenity_Internet",
    "Amenity_Closets",
    "Amenity_Ensuite",
    "Amenity_Annexe (Boys' quarters)",
    "Amenity_Gym",
]

amenitySum = pd.DataFrame()

for col in cols:
    counts = housing[col].value_counts().sort_index()
    percentage = (counts[1] / len(housing) * 100) if 1 in counts else 0

    amenitySum.loc[col, "1s"] = counts.get(1, 0)
    amenitySum.loc[col, "0"] = counts.get(0, 0)
    amenitySum.loc[col, "% 1s"] = f"{percentage:.2f}%"

amenitySum = amenitySum.sort_values("1s", ascending=False)
display(amenitySum)

# realised amenities were the most in the features so i wanted to check the proportions of these values and find potential issues that may arise when i use any
```

```python
housing = housing.drop(
    [
        "Amenity_Closets",
        "Amenity_Kitchen cabinets",
        "Amenity_Kitchen",
        "Amenity_Fully fenced",
        "Amenity_Ensuite",
        "Amenity_Intercom",
        "Amenity_Balcony",
        "Amenity_Outdoor entertaining area",
        "Amenity_Gym",
        "Amenity_Alarm system",
        "Amenity_Oven",
        "Amenity_Dishwasher",
        "Amenity_Carport",
        "Amenity_Pay TV access",
        
    ],
    axis=1,
)

# These ones had little to no observations hence adding them would be noise, since their coefficients will always be close to 0, tree models don't actually need you to do this since it handles that internally, but linear models need this to be done"
```

```python
groups = {
    "comfortFeatures": [
        "Amenity_Air conditioning",
        "Amenity_Fans",
        "Amenity_Washing machine",
        "Amenity_Microwave",
        "Amenity_Refrigerator",
        "Amenity_Cooker",
    ],
    "utilityFeatures": [
        "Amenity_Water reservoir",
        "Amenity_Generator",
        "Amenity_Security service",
        "Amenity_Garage",
    ],
    "connectivityFeatures": ["Amenity_Internet", "Amenity_Dstv"],
    "luxuryFeatures": [
        "Amenity_Garden",
        "Amenity_Swimming pool",
        "Amenity_Roof terrace",
    ],
    "exSpace": ["Amenity_Annexe (Boys' quarters)"],
}

for groupName, cols in groups.items():
    housing[groupName] = housing[cols].sum(
        axis=1
    )  # This function is just like count the number of features each listing have that falls under a particular group


# Some of the features despite had many obsevations were still imbalanced, so i thought that grouping some under some class could lead to more realised effect, it turned out that i was actually right, we'll see that in the next cells. Also, linear models would have very much noise if you feed in data with multicollinearity, this would be an issue hence grouping could solve that. Another reason is that linear models tend to perform poorly with too much cat. features, grouping would actually eliminate some
```

```python
corrMatrix = housing.corr()
corrMatrix['LogPrice'].sort_values(ascending=False)

# This shows that grouping them actually helps,, because you could see our engineered features tend to have much influence on the target than most of the rest if they would have been left alone"
```
